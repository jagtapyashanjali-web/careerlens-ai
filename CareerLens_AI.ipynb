```python
# Step 8: Extract skills from job descriptions

def extract_skills(text):
    found_skills = []
    for skill in skills:
        if skill.lower() in text.lower():
            found_skills.append(skill)
    return found_skills

jobs_df["detected_skills"] = jobs_df["job_description"].apply(extract_skills)

print("Skills extracted successfully!")

jobs_df[["job_title", "detected_skills"]]
```

```python
# Step 9: Count skill demand

from collections import Counter

skill_counts = Counter()

for job_skills in jobs_df["detected_skills"]:
    for skill in job_skills:
            skill_counts[skill] += 1

            skill_demand = pd.DataFrame(skill_counts.items(), columns=["Skill", "Number_of_Jobs"]).sort_values("Number_of_Jobs", ascending=False)

            print("skill demand calculated")
```

```python
from collections import Counter

skills = ["Python", "Java", "C", "C++", "JavaScript", "SQL", "HTML", "CSS", "React", "Node.js", "Django", "Flask", "Machine Learning", "Deep Learning", "Data Analysis", "Pandas", "NumPy", "Scikit-learn", "TensorFlow", "PyTorch", "Excel", "Power BI", "Git", "GitHub", "AWS", "Docker", "Linux", "Windows", "Networking", "Troubleshooting", "Ticketing", "Customer Support", "MS Office", "REST API", "Selenium"]

extract_skills = lambda text: [s for s in skills if s.lower() in text.lower()]
```

```python
# Step 47: Skill Demand in Expanded Job Dataset

real_jobs_df["detected_skills"] = real_jobs_df["job_description"].apply(extract_skills)

expanded_skill_counts = Counter()

for job_skills in real_jobs_df["detected_skills"]:
 for skill in job_skills:
                expanded_skill_counts[skill] += 1

                expanded_demand_df = pd.DataFrame(
                    expanded_skill_counts.items(),
                        columns=["Skill", "Job Count"]
                        ).sort_values(
                            "Job Count",
                                ascending=False
                                ).reset_index(drop=True)

print("=" * 60)
print("CAREERLENS AI - EXPANDED SKILL DEMAND")
print("=" * 60)

expanded_demand_df
```

```python
# Step 48: Expanded Skill Demand Visualization

top_expanded_skills = expanded_demand_df.head(15)

plt.figure(figsize=(10, 6))

plt.barh(
    top_expanded_skills["Skill"],
        top_expanded_skills["Job Count"]
        )

plt.xlabel("Number of Job Listings")
plt.ylabel("Skill")
plt.title("CareerLens AI - Most In-Demand Skills")

plt.gca().invert_yaxis()
plt.tight_layout()

plt.show()
```

```python
# Step 49: Skills Required by Job Role
role_skill_summary = real_jobs_df.explode("detected_skills").groupby(["job_title", "detected_skills"]).size().reset_index(name="Listings")

print("=" * 65)
print("CAREERLENS AI - SKILLS BY CAREER ROLE")
print("=" * 65)

role_skill_summary.sort_values(["job_title", "Listings"], ascending=[True, False])
```

```python
# Step 50: Career x Skill Summary
top5 = role_skill_summary.sort_values("Listings", ascending=False).groupby("job_title").head(5)

career_skill_df = top5.groupby("job_title")["detected_skills"].apply(", ".join).reset_index()
career_skill_df.columns = ["Career Role", "Top Skills"]

print("=" * 70)
print("           CAREERLENS AI - CAREER SKILL MAP")
print("=" * 70)

career_skill_df
```

```python
resume_text = "I am a BSc Computer Science graduate with skills in Python, Java, C, C++, SQL, HTML, CSS, MS Office and database management. I completed an Airbnb NYC data analysis project using Python, Pandas, Seaborn and Google Colab."
```

```python
# Step 52: Recalculate Resume Skills

detected_resume_skills = extract_skills(resume_text)

print("=" * 60)
print("CAREERLENS AI - UPDATED RESUME SKILLS")
print("=" * 60)

for skill in detected_resume_skills:
    print("✓", skill)

    print("\nTotal skills detected:", len(detected_resume_skills))
```

```python
# Step 53: Updated Skill Gap Analysis

target_job = real_jobs_df[
    real_jobs_df["job_title"] == target_role
    ].iloc[0]

required_skills = set(target_job["detected_skills"])
your_skills = set(detected_resume_skills)

matched_skills = required_skills.intersection(your_skills)
missing_skills = required_skills.difference(your_skills)

match_percentage = (
        len(matched_skills) / len(required_skills) * 100
            if required_skills else 0
            )

print("=" * 60)
print("CAREERLENS AI - UPDATED SKILL GAP")
print("=" * 60)

print("\nTarget Role:", target_role)
print("Skill Match:", round(match_percentage, 2), "%")

print("\n✅ Skills You Have:")
for skill in sorted(matched_skills):
    print("✓", skill)

print("\n⚠️ Skills Missing:")
for skill in sorted(missing_skills):
    print("→", skill)
```
